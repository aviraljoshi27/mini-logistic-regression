### NumPy Broadcasting Rules Reference

Broadcasting allows arithmetic operations on arrays of different shapes. For an operation to work, the dimensions of the two arrays must be compatible. 

**The Alignment Rule:** Always align shapes starting from the **right side** (trailing dimensions first) [2.1]. If one array has fewer dimensions, NumPy automatically prepends `1`s to its shape to match the length of the larger array [2.1].

| Shape A | Shape B | Expanded B | Rows Comparison | Columns Comparison | Status & Behavior |
| :--- | :--- | :--- | :--- | :--- | :--- |
| `(2, 3)` | `(2, 3)` | `(2, 3)` | Equal ✓ | Equal ✓ | **Works** (Standard element-by-element operation) |
| `(2, 3)` | `(1, 3)` | `(1, 3)` | 1 → Stretch ✓ | Equal ✓ | **Works** (Shape B stretches rows to match A) |
| `(2, 3)` | `(2, 1)` | `(2, 1)` | Equal ✓ | 1 → Stretch ✓ | **Works** (Shape B stretches columns to match A) |
| `(2, 3)` | `(1, 1)` | `(1, 1)` | 1 → Stretch ✓ | 1 → Stretch ✓ | **Works** (Acts like adding a single scalar) |
| `(2, 3)` | `(3,)` | `(1, 3)` | 1 → Stretch ✓ | Equal ✓ | **Works** ✓ (B aligns right, stretches across rows) [2.1] |
| `(2, 3)` | `(2,)` | `(1, 2)` | Equal ✓ | 3 vs 2 ✗ | ❌ **Error** (Trailing dimensions `3` and `2` mismatch) [2.1] |
| `(2, 3)` | `(2, 2)` | `(2, 2)` | Equal ✓ | 3 vs 2 ✗ | ❌ **Error** (Dimension mismatch; cannot stretch) |

---

### Core Takeaways

* **Right-to-Left Alignment:** Before comparing, NumPy pads the smaller shape on the *left* with `1`s [2.1]. This is why `(3,)` becomes `(1, 3)` (compatible) [2.1], but `(2,)` becomes `(1, 2)` (incompatible) [2.1].
* **The Rule of 1:** A dimension of size `1` acts like a rubber band. NumPy will automatically stretch/copy it to match the larger shape along that axis without using extra memory.
* **The Mismatch Rule:** If two aligned dimensions are different and neither is `1`, broadcasting fails, throwing a `ValueError` [2.1].

### Quick Fix for the `(2,)` Mismatch
To make a `(2,)` array work with a `(2, 3)` array, you must manually transform its shape from a row vector into a column vector `(2, 1)` [2.1] so that the trailing dimension is a stretchable `1`:
```python
# Convert shape (2,) to (2, 1) to force broadcasting across columns
array_B[:, np.newaxis]  # or array_B.reshape(-1, 1)
```

#### The Number `e`

The number **\(e\)** is a special mathematical constant, much like \(\pi\), roughly equal to **\(2.718\)**. It appears naturally in patterns of growth and probability, which is exactly why the formula for **logistic regression** is built on top of it. 

The expression **\(e^z\)** means *"\(e\) raised to the power of \(z\)"*.

> **\(e^z\) is always positive.** 
> Even for highly negative values of \(z\), the result only gets closer and closer to \(0\). It never actually reaches \(0\), and it never drops below it into negative numbers.

> **It grows extremely fast.** 
> Each \(+1\) added to \(z\) multiplies the previous result by roughly \(2.7\). By the time \(z = 10\), the value has already skyrocketed to **\(\approx 22,026\)**.


````python
# np.exp

print(np.e)
print(np.exp(z))

z_values = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])
print(np.exp(z_values))
````

## When `np.exp` overflows

The exponential function grows **very quickly**:

\[
e^z
\]

Each time we increase `z` by `1`, the value of \(e^z\) is multiplied by approximately **2.718**.

For example:

- \(e^1 \approx 2.718\)
- \(e^2 \approx 7.389\)
- \(e^{10} \approx 22,026\)
- \(e^{100} \approx 2.69 \times 10^{43}\)

### What is overflow?

A NumPy `float64` can store values only up to approximately:

\[
1.8 \times 10^{308}
\]

If `np.exp(z)` becomes larger than this limit, there is no room to store the result.

This is called **overflow**.

```python
np.exp(1000)
# inf

````python
big_z = np.array([100.0, 700.0, 710.0, 1000.0])
print(np.exp(big_z))

print(np.exp(-1000.0))
````

## The logarithm and `np.log`

### What is a logarithm?

A **logarithm** undoes a power.

The **natural logarithm**, written as `ln`, undoes exponentiation with the base \(e\).

In other words:

\[
\ln(x) = \text{"e to what power gives } x\text{"?"}
\]

For example:

\[
e^2 \approx 7.389
\]

Therefore:

\[
\ln(7.389) \approx 2
\]

So:

```python
np.log(7.389)
# ≈ 2

````python
x_values = np.array([np.e, 1.0, 0.5, 0.01, 0.0, -1.0])
print(np.log(x_values))

print(np.log(np.exp(3.0)))
````

## Clipping

### The concept

**Clipping** means forcing every number to stay between a **lower limit** and an **upper limit**.

The rule is simple:

- If a value is **below the lower limit** → raise it to the lower limit.
- If a value is **above the upper limit** → lower it to the upper limit.
- If a value is **between the limits** → leave it unchanged.

### Example

Suppose we want every value to stay between `0` and `10`:

| Original | Clipped |
|---:|---:|
| `-5` | `0` |
| `0` | `0` |
| `3` | `3` |
| `10` | `10` |
| `15` | `10` |

So clipping doesn't change values that are already inside the allowed range.

### In NumPy

NumPy provides `np.clip()` for this:

```python
np.clip(x, lower_limit, upper_limit)

````python
eps = 1e-15
# eps = 1e-15 is 0.000000000000001. eps is short for epsilon

probs = np.array([0.0, 0.3, 1.0])

safe_probs = np.clip(probs, eps, 1 - eps)
print(safe_probs)
print(np.log(safe_probs))
````
