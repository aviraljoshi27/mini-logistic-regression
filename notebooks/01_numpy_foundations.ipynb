```python
#Vectors and matrices (the concept)
X = np.array([
    [14.1, 20.3, 92.0],
    [11.4, 15.7, 73.2],
    [19.8, 22.1, 130.5],
    [12.5, 18.9, 80.1],
])

print(X)
print(X.shape)
print(X.ndim)
print(X.dtype)
```

```python
# Indexing a matrix (rows, columns, single values)

print(X[0])
print(X[:, 0])
print(X[2, 2])


print(X[0].shape, X[:, 0].shape)
```

```python
print(X[:2, :1])
print(X[:2, 1])

print(X[:2, :1].shape, X[:2, 1].shape)
```

```python
# Broadcasting (the concept)

marks = np.array([
    [10, 20, 30],
    [40, 50, 60],
])
bonus = np.array([1, 2, 3])

print(marks.shape, bonus.shape)
print(marks - bonus)
print((marks - bonus).shape)
```

### 📐 NumPy Broadcasting Rules Reference

Broadcasting allows arithmetic operations on arrays of different shapes. For an operation to work, the dimensions of the two arrays must be compatible. Starting from the trailing (rightmost) dimensions, axes are compatible if they are **equal**, or if **one of them is 1**.

| Shape A | Shape B | Rows Comparison | Columns Comparison | Status & Behavior |
| :--- | :--- | :--- | :--- | :--- |
| `(2, 3)` | `(2, 3)` | Equal ✓ | Equal ✓ | **Works** (Standard element-by-element operation) |
| `(2, 3)` | `(1, 3)` | 1 → Stretch ✓ | Equal ✓ | **Works** (Shape B stretches rows to match A) |
| `(2, 3)` | `(2, 1)` | Equal ✓ | 1 → Stretch ✓ | **Works** (Shape B stretches columns to match A) |
| `(2, 3)` | `(1, 1)` | 1 → Stretch ✓ | 1 → Stretch ✓ | **Works** (Acts like adding a single scalar) |
| `(2, 3)` | `(2, 2)` | Equal ✓ | 3 vs 2 ✗ | ❌ **Error** (Dimension mismatch; cannot stretch) |

---

### 💡 Core Takeaways

* **The Rule of 1:** A dimension of size `1` acts like a rubber band. NumPy will automatically stretch/copy it to match the larger shape along that axis without using extra memory.
* **The Mismatch Rule:** If two dimensions are different and neither is `1` (like `3` vs `2` in the final row), broadcasting fails, throwing a `ValueError`.
* **Direction of Alignment:** Always align shapes starting from the **right side** (columns first, then rows). If one array has fewer dimensions, NumPy prepends `1`s to its shape.
